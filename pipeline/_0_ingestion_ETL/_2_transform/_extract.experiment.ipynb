## Dropping PII cols and irrelevant cols

Within the dataset, there PII columns in there, specifically, `applicant details`, `recruiters`. Since we don't want this senstive data in our data warehouse, we're going to remove it.

We are only going to leave the applicant id for identity of a given application. All of the recruiters detail, however, will be removed, since it gives no analytical insigts later on, not unless, we are creating an internal level to give insights about employee performance. Since we are not, we are going to remove such column. 

On top of this, the `notes` column also yields no analytical insight either. Since this will be irrelevant now, we will also remove this one.


```python
cols_to_remove = ["applicant_name", "email", "recruiter","notes"]
df_safe = df_.drop(cols_to_remove)

df_safe.head(2)
```
